```python
corpus="""Hello welcome to krish naik's NLP tutorials.
Please do watch the entire course to become an expert in NLP
            
"""
```

```python
print(corpus)
```

```python
# Tokenization
# sentence --> paragraph
from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
# Tokenization 
# Paragraph --> words
# sentence --> words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)  # punctuations are considered as separate word
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()  # full stop will not be considered as separate word
```

```python
tokenizer.tokenize(corpus)
```
